```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set a style for better aesthetics
sns.set_style("whitegrid")

# Re-create the datasets for standalone execution
np.random.seed(42)

# Simulated Churn Dataset
n_samples = 1000
age = np.random.randint(18, 65, n_samples)
salary = np.random.randint(30000, 120000, n_samples)
tenure = np.random.randint(1, 10, n_samples) # Years with company
product_type = np.random.choice(['A', 'B', 'C', 'D'], n_samples)
churn_prob = (
    0.05 * (65 - age) / 47 +
    0.05 * (120000 - salary) / 90000 +
    0.1 * (10 - tenure) / 9
)
churn_prob += np.where(product_type == 'D', 0.2, 0)
churn_prob = np.clip(churn_prob + np.random.normal(0, 0.1, n_samples), 0, 1)
churn = (churn_prob > 0.5).astype(int)

df_churn = pd.DataFrame({
    'Age': age,
    'Salary': salary,
    'Tenure': tenure,
    'Product_Type': product_type,
    'Churn': churn
})
df_churn['Churn'] = df_churn['Churn'].map({0: 'No Churn', 1: 'Churn'}) # Make target readable


# Simulated Housing Dataset
n_samples_reg = 500
sq_footage = np.random.randint(800, 3000, n_samples_reg)
num_bedrooms = np.random.randint(2, 6, n_samples_reg)
location_score = np.random.rand(n_samples_reg) * 10
price = (sq_footage * 100 + num_bedrooms * 5000 + location_score * 10000 +
         np.random.normal(0, 50000, n_samples_reg))
price = np.maximum(0, price)

df_housing = pd.DataFrame({
    'Sq_Footage': sq_footage,
    'Num_Bedrooms': num_bedrooms,
    'Location_Score': location_score,
    'Price': price
})

print("--- Visualization Hands-on Exercises ---")


# --- Exercise 1: Histograms and KDE Plots for Distributions ---

print("\n--- Exercise 1: Histograms and KDE Plots ---")
plt.figure(figsize=(12, 5))

# 1. Histogram of 'Salary' in df_churn
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
sns.histplot(df_churn['Salary'], bins=30, kde=True, color='skyblue')
plt.title('Distribution of Customer Salary')
plt.xlabel('Salary')
plt.ylabel('Count / Density')
```

```python

# 2. KDE plot of 'Tenure' in df_churn, separated by 'Churn' status
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
sns.kdeplot(data=df_churn, x='Tenure', hue='Churn', fill=True, palette='viridis')
plt.title('Tenure Distribution by Churn Status')
plt.xlabel('Tenure (Years)')
plt.ylabel('Density')
plt.tight_layout() # Adjust layout to prevent overlapping
plt.show()
print("-" * 50)
```

```python

# --- Exercise 2: Scatter Plots for Relationships ---

print("\n--- Exercise 2: Scatter Plots ---")
plt.figure(figsize=(14, 6))

# 1. Scatter plot of 'Age' vs 'Salary' in df_churn, colored by 'Churn' status
plt.subplot(1, 2, 1)
sns.scatterplot(x='Age', y='Salary', hue='Churn', data=df_churn, alpha=0.7, s=50, palette='coolwarm')
plt.title('Age vs. Salary by Churn Status')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.legend(title='Churn Status')
```

```python

# 2. Scatter plot of 'Sq_Footage' vs 'Price' in df_housing, with 'Num_Bedrooms' influencing point size.
plt.subplot(1, 2, 2)
sns.scatterplot(x='Sq_Footage', y='Price', size='Num_Bedrooms', data=df_housing,
                hue='Num_Bedrooms', sizes=(50, 500), alpha=0.6, palette='magma')
plt.title('House Price vs. Square Footage by Number of Bedrooms')
plt.xlabel('Square Footage')
plt.ylabel('Price')
plt.legend(title='Num Bedrooms')
plt.tight_layout()
plt.show()
print("-" * 50)
```

```python

# --- Exercise 3: Bar Plots and Count Plots for Categorical Data ---

print("\n--- Exercise 3: Bar Plots and Count Plots ---")
plt.figure(figsize=(14, 6))

# 1. Count plot of 'Product_Type' in df_churn
plt.subplot(1, 2, 1)
sns.countplot(x='Product_Type', data=df_churn, palette='pastel')
plt.title('Count of Customers per Product Type')
plt.xlabel('Product Type')
plt.ylabel('Number of Customers')

# 2. Bar plot showing the average 'Salary' per 'Product_Type', separated by 'Churn' status.
plt.subplot(1, 2, 2)
sns.barplot(x='Product_Type', y='Salary', hue='Churn', data=df_churn, palette='rocket')
plt.title('Average Salary by Product Type and Churn Status')
plt.xlabel('Product Type')
plt.ylabel('Average Salary')
plt.tight_layout()
plt.show()
print("-" * 50)
```

```python

# --- Exercise 4: Box Plots for Distribution by Category ---

print("\n--- Exercise 4: Box Plots ---")
plt.figure(figsize=(10, 6))
# 1. Box plot of 'Sales' (from a hypothetical 'df_sales' column or df_housing prices if we rename)
#    Let's use 'Price' from df_housing and categorize it for demonstration.
df_housing['Price_Category'] = pd.qcut(df_housing['Price'], q=3, labels=['Low', 'Medium', 'High'])
sns.boxplot(x='Price_Category', y='Sq_Footage', data=df_housing, palette='viridis')
plt.title('Square Footage Distribution by Price Category')
plt.xlabel('Price Category')
plt.ylabel('Square Footage')
plt.show()
print("-" * 50)


# --- Exercise 5: Correlation Heatmap (for numerical features) ---

print("\n--- Exercise 5: Correlation Heatmap ---")
# 1. Calculate the correlation matrix for numerical features in df_housing.
numerical_housing_features = ['Sq_Footage', 'Num_Bedrooms', 'Location_Score', 'Price']
correlation_matrix = df_housing[numerical_housing_features].corr()

plt.figure(figsize=(8, 7))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Housing Features')
plt.show()
print("-" * 50)


# --- Exercise 6: Faceting / Subplots with Seaborn (Advanced Visualization) ---
# Useful for comparing distributions or relationships across different subsets of data

print("\n--- Exercise 6: Faceting / Subplots ---")
# 1. Use `sns.FacetGrid` to visualize the relationship between 'Age' and 'Salary'
#    for each 'Product_Type' and 'Churn' status.
g = sns.FacetGrid(df_churn, col="Product_Type", row="Churn", hue="Churn",
                  palette="coolwarm", margin_titles=True, height=3, aspect=1.2)
g.map(sns.scatterplot, "Age", "Salary", alpha=0.7, s=50)
g.add_legend()
g.set_axis_labels("Age", "Salary")
g.set_titles(col_template="Product Type: {col_name}", row_template="Churn: {row_name}")
plt.suptitle('Age vs. Salary by Product Type and Churn Status', y=1.02) # Adjust suptitle position
plt.tight_layout()
plt.show()
print("-" * 50)
```
